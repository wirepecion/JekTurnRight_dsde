```python
import pandas as pd
df = pd.read_csv('example.csv')
df
```

```python
# make year month day columns to datetime
# rename to expected keys (year, month, day) before assembling
df['date'] = pd.to_datetime(
	df[['year_timestamp', 'month_timestamp', 'days_timestamp']].rename(
		columns={'year_timestamp': 'year', 'month_timestamp': 'month', 'days_timestamp': 'day'}
	)
)

# drop year month day columns
df = df.drop(columns=['year_timestamp', 'month_timestamp', 'days_timestamp'])
df
```

```python
df.to_csv('example_with_date.csv', index=False)
```
